```python
crocs2 = crocs.query('Genus == "Crocodylus"')
crocs2.sample(3)
```

**Gráfica 1**

```python
px.scatter(crocs2, x='Length_m', y='Weight_kg',
           title='Dimensiones de cocodrilos: largo vs Peso') # simple
```

**Gráfica 2**

```python
px.histogram(crocs2, x='Country/Region') # sencillo
```

**Gráfica 3**

```python
px.bar(crocs2, x='Country/Region', y='Length_m')
```

**Gráfica 4**

```python
px.bar(crocs2, x='Habitat Type', y='Length_m',
       hover_name='Scientific Name', hover_data=['Country/Region'],
       color='Conservation Status')
```

**Gráfica 5**

```python
# Identificar path: asociacion/camino entre jerarquias
# Ejemplo: continente -> pais
px.sunburst(crocs2, path=['Country/Region', 'Conservation Status']) # simple, por conteo
```

**Gráfica 6**

```python
px.histogram(crocs2, x='Habitat Type') # sencillo
```

**Gráfica 7**

```python
px.histogram(crocs2, x='Habitat Type', color='Conservation Status',
            title='Distribución de tipos e habitat por estado de conservación') # stacked histograms
```

**Gráfica 8**

```python
fig = px.histogram(crocs2,
                   x='Age Class',
                   color='Age Class',
                   title='Conteo de clase por edad')
fig.show()

#px.bar(crocs2, x='Age Class')
```
